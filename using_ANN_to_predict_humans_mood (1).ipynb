```python
import pandas as pd
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
# Load dataset
data = pd.read_csv("Sleep_health_and_lifestyle_dataset.csv")
data
```

```python
# Remove extra spaces
data.columns = data.columns.str.strip()
```

```python
# Select needed columns
X = data[['Sleep Duration', 'Stress Level', 'Physical Activity Level']].values
```

```python
# Convert Stress Level → Mood
# Low stress = Happy (0)
# Medium stress = Tired (2)
# High stress = Stress (1)

def convert_mood(stress):
    if stress <= 3:
        return 0   # Happy
    elif stress <= 6:
        return 2   # Tired
    else:
        return 1   # Stress

y = data['Stress Level'].apply(convert_mood).values
```

```python
# Create ANN model
model = Sequential()
model.add(Dense(8, input_dim=3, activation='relu'))
model.add(Dense(3, activation='softmax'))
```

```python
# Compile model
model.compile(loss='sparse_categorical_crossentropy',
              optimizer='adam',
              metrics=['accuracy'])
```

```python
# Train model
model.fit(X, y, epochs=50)
```

```python
# Take user input
sleep = float(input("Enter Sleep Hours: "))
stress = float(input("Enter Stress Level (1-10): "))
activity = float(input("Enter Physical Activity Level: "))
```

```python
# Convert into numpy array
test = np.array([[sleep, stress, activity]])
pred = model.predict(test)
labels = ["Happy", "Stress", "Tired"]
print("Mood:", labels[np.argmax(pred)])
```
